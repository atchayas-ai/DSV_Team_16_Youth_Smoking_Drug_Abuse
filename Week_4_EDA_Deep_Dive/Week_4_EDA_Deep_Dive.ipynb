## 4.0 Importing Libraries and Loading the Cleaned Dataset

Run this setup section first. Week 4 uses the cleaned dataset created in Week 3. Because this notebook is inside the Week 4 folder, the relative path points to the `Dataset` folder at the repository root.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

cleaned_path = Path('../Dataset/youth_drug_consumption_cleaned.csv')

if not cleaned_path.exists():
    raise FileNotFoundError(
        f'Cleaned dataset not found at {cleaned_path}. '
        'Run the Week 3 notebook and save the cleaned CSV in the Dataset folder.'
    )

df_clean = pd.read_csv(cleaned_path)
df_eda = df_clean.copy()

personality_cols = ['Nscore', 'Escore', 'Oscore', 'Ascore', 'Cscore', 'Impulsive', 'SS']

print('Cleaned dataset loaded successfully!')
print('Dataset shape:', df_clean.shape)
display(df_clean.head())
```

## Data Preparation for EDA

Some demographic variables in the dataset are represented using numerical codes. To make the exploratory analysis easier to understand, the encoded Gender values are converted into meaningful labels such as Male and Female. The original encoded Gender column is retained, and a new labelled column is created for visualization.

```python
# Create a copy of the cleaned dataset for EDA
df_eda = df_clean.copy()

# Convert encoded Gender values into readable labels
df_eda["Gender_Label"] = df_eda["Gender"].map({
    -0.48246: "Male",
     0.48246: "Female"
})

# Display original and converted values
df_eda[["Gender", "Gender_Label"]].head(10)
```

### Checking the Gender Categories

The converted Gender labels are checked to ensure that the numerical codes have been correctly represented as meaningful categories before visualization.

```python
df_eda["Gender_Label"].value_counts()
```

## 4.1 Univariate Analysis — Gender

Gender is a categorical variable with two categories: Male and Female.
A count plot is used to visualize the number of observations in each gender category.

```python
plt.figure(figsize=(7, 5))

sns.countplot(x="Gender_Label", data=df_eda)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

## Data Preparation — Age

The Age variable is represented using numerical codes corresponding to different age groups. For easier interpretation and visualization, the encoded values are converted into meaningful age-group labels. The original Age column is retained, and a new Age_Group column is created.

```python
# Convert encoded Age values into meaningful age groups

df_eda["Age_Group"] = df_eda["Age"].map({
    -0.95197: "18-24",
    -0.07854: "25-34",
     0.49788: "35-44",
     1.09449: "45-54",
     1.82213: "55-64",
     2.59171: "65+"
})

# Display original and converted values
df_eda[["Age", "Age_Group"]].head(10)
```

### Checking the Age Groups

The newly created Age_Group column is checked to verify that the encoded Age values have been correctly converted into meaningful age categories.

```python
df_eda["Age_Group"].value_counts()
```

## 4.2 Univariate Analysis — Age Group

Age is represented using encoded age groups in the dataset.
A count plot is used to visualize the distribution of observations across different age groups.

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    x="Age_Group",
    data=df_eda,
    order=["18-24", "25-34", "35-44", "45-54", "55-64", "65+"]
)

plt.title("Age Group Distribution")
plt.xlabel("Age Group")
plt.ylabel("Count")
plt.show()
```

## Data Preparation — Education

The Education variable is represented using numerical codes corresponding to different education levels. To make the variable easier to interpret during exploratory analysis, the encoded values are converted into meaningful education-level labels while retaining the original column.

```python
df_eda = df_clean.copy()

df_eda["Education_Level"] = df_eda["Education"].map({
    -2.43591: "Left School Before 16",
    -1.73790: "Left School at 16",
    -1.43719: "Left School at 17",
    -1.22751: "Left School at 18",
    -0.61113: "Some College/University",
     0.45468: "Professional Certificate/Diploma",
     1.16365: "University Degree",
     1.98437: "Masters Degree",
     2.12700: "Doctorate Degree",
    -0.05921: "Median Imputed"
})

df_eda[["Education", "Education_Level"]].head(10)
```

### 4.3 Univariate Analysis - Education Level Distribution

Education is an ordinal categorical variable represented using numerical codes.
A count plot is used to visualize the number of observations in each education level.

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    y="Education_Level",
    data=df_eda
)

plt.title("Distribution of Education Levels")
plt.xlabel("Count")
plt.ylabel("Education Level")
plt.show()
```

### 4.4 Univariate Analysis – Personality Score Distributions

The personality variables are numerical variables that represent different personality traits.
Histograms are used to visualize their distributions and understand their spread and concentration.

```python
personality_cols = [
    "Nscore",
    "Escore",
    "Oscore",
    "Ascore",
    "Cscore",
    "Impulsive",
    "SS"
]

df_eda[personality_cols].hist(
    figsize=(12, 8),
    bins=15
)

plt.suptitle("Distribution of Personality Scores")
plt.show()
```

### 4.5 Univariate Analysis – Drug Consumption Distribution

Drug consumption variables are categorical variables represented by consumption levels from CL0 to CL6.
Count plots are used to visualize the frequency of each consumption level for different substances.

```python
drug_cols = [
    "Alcohol",
    "Cannabis",
    "Coke",
    "Heroin",
    "Ecstasy",
    "LSD",
    "Meth",
    "Mushrooms",
    "Nicotine"
]

for drug in drug_cols:
    plt.figure(figsize=(7, 4))
    sns.countplot(x=drug, data=df_eda)
    plt.title(f"{drug} Consumption Distribution")
    plt.xlabel("Consumption Level")
    plt.ylabel("Count")
    plt.show()
```

### Univariate Analysis – Conclusion

The univariate analysis helped us understand the distribution of individual variables in the dataset.
The categorical variables such as gender, age group, education level, and drug consumption were analyzed using count plots, while personality scores were analyzed using histograms.
The plots helped identify the frequency, distribution, and concentration of different values in the dataset.

### 4.14 Correlation Heatmap of Personality Scores

A correlation heatmap is used to examine the strength and direction of relationships among numerical personality variables.
Correlation values range from -1 to +1, where values closer to +1 indicate a positive relationship and values closer to -1 indicate a negative relationship.

```python
correlation = df_eda[personality_cols].corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
                )

plt.title("Correlation Heatmap of Personality Scores")
plt.show()
```
