## 2.7 Missing Value Check

Missing values are checked in each column to identify incomplete observations. This step helps determine which variables require attention during the data-cleaning process.

```python
missing_values = df.isnull().sum()

missing_values
```

### Missing Value Summary

Only the variables containing missing values are displayed to make the data-quality issues easier to identify.

```python
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df) * 100).round(2)
})

missing_summary = missing_summary[missing_summary["Missing Values"] > 0]

missing_summary
```

## 2.8 Duplicate Value Check

Duplicate records are checked to determine whether identical rows are present in the dataset. Duplicate observations can affect the accuracy of statistical analysis and should be addressed during data cleaning if found.

```python
duplicate_count = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_count)
```

## 2.9 Unique Value Inspection

The number of unique values in each column is examined to understand the variety of values and identify categorical variables that may require special treatment during data cleaning.

```python
unique_values = pd.DataFrame({
    "Column": df.columns,
    "Unique Values": df.nunique()
})

unique_values
```

### Drug Consumption Categories

The unique categories of selected drug-consumption variables are inspected to understand how consumption levels are represented in the dataset.

```python
print("Alcohol categories:")
print(df["Alcohol"].unique())

print("\nCannabis categories:")
print(df["Cannabis"].unique())

print("\nCocaine categories:")
print(df["Coke"].unique())
```

## 2.10 First Impression Report

The first impression report summarizes the initial findings obtained from the dataset inspection. It provides an overview of the dataset size, variable types, missing values, and duplicate records before data cleaning is performed.

```python
print("========== FIRST IMPRESSION REPORT ==========")

print("Dataset Name        : Drug Consumption UCI - Missing Values")
print("Number of Rows      :", df.shape[0])
print("Number of Columns   :", df.shape[1])

print("\nData Types:")
print(df.dtypes.value_counts())

print("\nMissing Value Columns:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nDuplicate Rows      :", df.duplicated().sum())

print("\nDataset contains both numerical and categorical variables.")
print("Further data cleaning is required because missing values are present.")
```
