## 3.1 Selecting Imputation Strategies

Age and Education are ordinal codes, Gender is binary categorical, and Nscore
is a continuous trait score. Median is used for Age, Education, and Nscore to
preserve their order/scale and stay robust to outliers. Mode is used for
Gender, since it's the only unordered categorical variable.


```python
df_clean = df.copy()
```

```python
for column in ["Age", "Gender", "Education", "Nscore"]:
      n_unique = df_clean[column].nunique()
      print(f"{column}: {n_unique} unique values")
```

## 3.2 Applying Imputation

Based on the strategy selected in the previous step, median imputation is
applied to `Age`, `Education`, and `Nscore`, while mode imputation is applied
to `Gender`. This replaces missing values with statistically appropriate
substitutes — the median for ordinal/continuous variables and the mode for
the binary categorical variable — without distorting the overall distribution
of each column.


```python
# Median imputation for numerical/ordinal variables
for column in ["Age", "Education", "Nscore"]:
    df_clean[column] = df_clean[column].fillna(
        df_clean[column].median()
    )

# Mode imputation for categorical variable
df_clean["Gender"] = df_clean["Gender"].fillna(
    df_clean["Gender"].mode()[0]
)

print("Missing values have been imputed.")
```

## 3.3 Verifying Imputation

After imputation, the dataset is checked again to confirm that the missing
values in Age, Gender, Education, and Nscore have been successfully handled.

```python
missing_after = df_clean.isnull().sum()
missing_after[missing_after > 0]
```

## 3.4 Preparing for Outlier Detection

Numerical variables are selected for outlier analysis because boxplots and
Z-scores are commonly used to identify unusually high or low numerical
observations.


```python
# Select continuous personality score variables
personality_cols = [
    "Nscore",
    "Escore",
    "Oscore",
    "Ascore",
    "Cscore",
    "Impulsive",
    "SS"
]

numeric_columns = personality_cols

print("Selected numerical variables:")
print(numeric_columns)
```

## 3.5 Outlier Detection Using Boxplots

Boxplots are used to visually identify potential outliers. Values located
beyond the whiskers of a boxplot may indicate observations that are unusually
high or low compared with the rest of the data.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

df_clean[numeric_columns].boxplot()

plt.xticks(rotation=90)
plt.title("Boxplot of Numerical Variables")
plt.ylabel("Values")
plt.show()
```

## 3.6 Outlier Detection Using Z-score

The Z-score method is used to identify numerical observations that are
unusually far from the mean. Observations with an absolute Z-score greater
than 3 are considered potential outliers for further investigation.

```python
from scipy.stats import zscore

z_scores = np.abs(zscore(df_clean[numeric_columns], nan_policy="omit"))

outlier_count = (z_scores > 3).sum(axis=0)

outlier_summary = pd.DataFrame({
    "Column": numeric_columns,
        "Outlier Count (|Z| > 3)": outlier_count
        })

print(outlier_summary)
```

## 3.7 Final Data Quality Check

The cleaned dataset is checked to confirm that missing values have been
handled and that the dataset retains its original number of observations and
variables. Potential outliers are reported separately rather than
automatically deleted.

```python
print("Final Shape:", df_clean.shape)

print("\nRemaining Missing Values:")
print(df_clean.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df_clean.duplicated().sum())
```

## 3.8 Cleaning Sprint — Conclusion

The dataset was cleaned by identifying and handling missing values in Age,
Gender, Education, and Nscore using appropriate imputation strategies. Median
imputation was applied to Age, Education, and Nscore, while mode imputation
was applied to Gender, based on the ordinal/continuous versus categorical
nature of each variable. Boxplots and Z-score analysis were then used to
identify potential outliers across all numerical variables. Rather than
automatically removing these outliers, they were reported for further
investigation, since extreme values in this dataset may represent legitimate
observations rather than data errors. The final dataset retains its original
1885 observations and 32 variables, with no remaining missing values and no
duplicate rows.

## 6.1 Identification of Numerical and Categorical Features

Features are identified based on their data types and meanings.

Numerical features contain measurable values, while categorical features represent groups or categories.

```python
personality_cols = [
    "Nscore", "Escore", "Oscore",
    "Ascore", "Cscore", "Impulsive", "SS"
]

drug_cols = [
    "Alcohol", "Amphet", "Amyl", "Benzos",
    "Caff", "Cannabis", "Choc", "Coke",
    "Crack", "Ecstasy", "Heroin", "Ketamine",
    "Legalh", "LSD", "Meth", "Mushrooms",
    "Nicotine", "Semer", "VSA"
]

print("Numerical Features:")
print(personality_cols)

print("\nCategorical Features:")
print(drug_cols)
```

## 6.2 Label Encoding

Label Encoding converts categorical values into numerical representations.

The drug consumption categories CL0 to CL6 are converted into numerical values while preserving their order.

```python
encoding_map = {
    "CL0": 0,
    "CL1": 1,
    "CL2": 2,
    "CL3": 3,
    "CL4": 4,
    "CL5": 5,
    "CL6": 6
}

df_label_encoded = df_fe.copy()

for column in drug_cols:
    df_label_encoded[column] = df_label_encoded[column].map(encoding_map)

print("Label Encoding Completed!")
display(df_label_encoded[drug_cols].head())
```

## 6.3 One-Hot Encoding

One-Hot Encoding converts categorical values into separate binary columns.

It is useful for categorical variables that do not have a natural numerical order.

```python
gender_data = df_fe.drop(columns=["Gender"]).copy()

gender_data["Gender_Label"] = df_fe["Gender"].map({
    -0.48246: "Male",
     0.48246: "Female"
})

df_onehot = pd.get_dummies(
    gender_data,
    columns=["Gender_Label"],
    dtype=int
)

print("One-Hot Encoding Completed!")
display(df_onehot.head())

print("New Shape:", df_onehot.shape)
```

## 6.4 Standardization

Standardization transforms numerical features so that they have approximately zero mean and unit standard deviation.

StandardScaler is used for the personality score variables.

```python
scaler = StandardScaler()

df_standard = df_fe.copy()

df_standard[personality_cols] = scaler.fit_transform(
    df_fe[personality_cols]
)

print("Standardization Completed!")
display(df_standard[personality_cols].head())

print("\nMean after standardization:")
print(df_standard[personality_cols].mean().round(2))

print("\nStandard deviation:")
print(df_standard[personality_cols].std().round(2))
```

```python
minmax_scaler = MinMaxScaler()

df_minmax = df_fe.copy()

df_minmax[personality_cols] = minmax_scaler.fit_transform(
    df_fe[personality_cols]
)

print("Min-Max Scaling Completed!")
display(df_minmax[personality_cols].head())
```

## 6.5 Skewness Analysis

Skewness measures the asymmetry of a data distribution.

- Positive skewness indicates a longer right tail.
- Negative skewness indicates a longer left tail.
- Skewness near zero indicates a relatively symmetrical distribution.

```python
skew_values = df_fe[personality_cols].skew()

print("Skewness Values:")
print(skew_values)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=skew_values.index,
    y=skew_values.values
)

plt.axhline(0, linestyle="--", color="black")
plt.title("Skewness of Personality Features")
plt.xlabel("Features")
plt.ylabel("Skewness")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 6.6 Skewness Correction

Yeo-Johnson transformation is used to reduce skewness in numerical data.

It can handle both positive and negative values.
Features with absolute skewness greater than 0.5 are selected for transformation.

```python
skewed_cols = skew_values[
    abs(skew_values) > 0.5
].index.tolist()

print("Selected Skewed Features:")
print(skewed_cols)

df_transformed = df_fe.copy()

if skewed_cols:
    transformer = PowerTransformer(
        method="yeo-johnson",
        standardize=False
    )

    df_transformed[skewed_cols] = transformer.fit_transform(
        df_fe[skewed_cols]
    )

    print("Yeo-Johnson Transformation Completed!")

else:
    print("No features selected for transformation.")

display(df_transformed[personality_cols].head())
```

## 6.7 Before and After Transformation Comparison

The skewness values before and after Yeo-Johnson transformation are compared to understand the effect of the transformation.

```python
before_skew = df_fe[personality_cols].skew()
after_skew = df_transformed[personality_cols].skew()

skew_comparison = pd.DataFrame({
    "Before Transformation": before_skew,
    "After Transformation": after_skew
})

display(skew_comparison)

skew_comparison.plot(
    kind="bar",
    figsize=(12, 5)
)

plt.title("Skewness Before and After Transformation")
plt.xlabel("Personality Features")
plt.ylabel("Skewness")
plt.axhline(0, color="black", linestyle="--")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 6.8 Binning and Discretization

Binning is the process of dividing continuous numerical data into intervals or groups.

Two methods are used:

1. Equal-width binning – divides the data into intervals of equal width.
2. Equal-frequency binning – divides the data into groups containing approximately equal numbers of observations.

```python
# Equal-width binning
df_binned = df_fe.copy()

df_binned["Nscore_EqualWidth"] = pd.cut(
    df_binned["Nscore"],
    bins=4,
    labels=["Low", "Medium", "High", "Very High"]
)

# Equal-frequency binning
df_binned["Nscore_EqualFrequency"] = pd.qcut(
    df_binned["Nscore"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)

print("Binning Completed!")

display(
    df_binned[
        ["Nscore", "Nscore_EqualWidth", "Nscore_EqualFrequency"]
    ].head(10)
)
```

# WEEK 7 – DIMENSIONALITY REDUCTION

## Aim
To reduce the dimensionality of the dataset while retaining important information using feature selection and Principal Component Analysis.

## Activities Performed
1. Feature Selection.
2. Variance Thresholding.
3. Correlation-Based Feature Selection.
4. Multicollinearity Analysis using VIF.
5. Principal Component Analysis (PCA).
6. Explained Variance Analysis.
7. PCA Visualization.

## Key Findings
- Seven personality features were selected.
- No features were removed through Variance Thresholding.
- No highly correlated feature pairs were found using the selected threshold of 0.80.
- VIF values were below 5.
- Six principal components retained approximately 95.07% of the variance.

## Outcome
The original seven features were reduced to six principal components while retaining approximately 95.07% of the variance.

```python
from sklearn.feature_selection import VarianceThreshold
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from statsmodels.stats.outliers_influence import variance_inflation_factor

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create a separate dataset
df_dr = df_clean.copy()

# Select personality features
features = [
    "Nscore", "Escore", "Oscore",
    "Ascore", "Cscore", "Impulsive", "SS"
]

X = df_dr[features].copy()

print("Dataset Shape:", X.shape)
print("\nSelected Features:")
print(features)

display(X.head())
```

## 7.1 Feature Selection

Feature Selection is the process of selecting the most relevant features from a dataset.

It helps remove irrelevant or less informative features, reduce computation time and simplify analysis.

```python
print("Selected Features:")
for feature in features:
    print(feature)

print("\nTotal Number of Features:", len(features))
```

## 7.2 Variance Thresholding

Variance Thresholding is a feature selection technique that removes features with very low variance.

Features with almost constant values provide less information for analysis.

A threshold of 0.01 is used to identify low-variance features.

```python
# Calculate variance
feature_variance = X.var()

print("Variance of Each Feature:")
print(feature_variance)

# Apply Variance Thresholding
selector = VarianceThreshold(threshold=0.01)

X_variance = selector.fit_transform(X)

selected_features = X.columns[
    selector.get_support()
].tolist()

removed_features = X.columns[
    ~selector.get_support()
].tolist()

print("\nRetained Features:")
print(selected_features)

print("\nRemoved Features:")
print(removed_features)

print("\nFeatures Before:", X.shape[1])
print("Features After:", len(selected_features))
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=feature_variance.index,
    y=feature_variance.values
)

plt.axhline(
    y=0.01,
    color="red",
    linestyle="--",
    label="Threshold = 0.01"
)

plt.title("Feature Variance Analysis")
plt.xlabel("Features")
plt.ylabel("Variance")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

## 7.3 Correlation-Based Feature Selection

Correlation-Based Feature Selection identifies relationships between numerical features.

Highly correlated features may contain overlapping information.

A correlation threshold of 0.80 is used to identify strongly correlated feature pairs.

```python
# Calculate correlation matrix
correlation_matrix = X.corr()

print("Correlation Matrix:")
display(correlation_matrix)

# Visualize correlation
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Personality Features Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Identify highly correlated feature pairs
threshold = 0.80

high_corr_pairs = []

for i in range(len(features)):
    for j in range(i + 1, len(features)):

        corr_value = correlation_matrix.iloc[i, j]

        if abs(corr_value) > threshold:
            high_corr_pairs.append(
                (features[i], features[j], corr_value)
            )

print("Highly Correlated Feature Pairs:")

if high_corr_pairs:
    for pair in high_corr_pairs:
        print(pair)
else:
    print("No highly correlated feature pairs found.")
```

## 7.4 Multicollinearity Analysis

Multicollinearity occurs when two or more independent features are highly correlated.

Variance Inflation Factor (VIF) is used to identify multicollinearity.

A VIF value greater than 5 is considered a warning sign that a feature needs further investigation.

```python
# Standardize features before VIF calculation
scaler_vif = StandardScaler()

X_vif = pd.DataFrame(
    scaler_vif.fit_transform(X),
    columns=features
)

# Calculate VIF
vif_data = pd.DataFrame()

vif_data["Feature"] = features

vif_data["VIF"] = [
    variance_inflation_factor(
        X_vif.values, i
    )
    for i in range(X_vif.shape[1])
]

print("Variance Inflation Factor:")
display(vif_data)

print("\nFeatures with VIF greater than 5:")

print(
    vif_data[vif_data["VIF"] > 5]
)
```

## 7.5 Principal Component Analysis (PCA)

Principal Component Analysis is a dimensionality reduction technique that transforms the original correlated features into a smaller set of uncorrelated variables called principal components.

PCA helps reduce the number of dimensions while retaining as much information as possible.

```python
# Standardize the original personality features
scaler_pca = StandardScaler()

X_scaled = scaler_pca.fit_transform(X)

# Apply PCA
pca = PCA()

X_pca = pca.fit_transform(X_scaled)

print("Original Number of Features:", X.shape[1])
print("Number of Principal Components:", X_pca.shape[1])

print("\nPCA Completed Successfully!")
```

## 7.6 Explained Variance Ratio

Explained Variance Ratio indicates how much information or variance each principal component retains from the original dataset.

Cumulative Explained Variance represents the total variance retained by the selected components.

```python
explained_variance = pca.explained_variance_ratio_

cumulative_variance = np.cumsum(
    explained_variance
)

variance_table = pd.DataFrame({
    "Principal Component": [
        f"PC{i+1}" for i in range(len(explained_variance))
    ],
    "Explained Variance": explained_variance,
    "Cumulative Variance": cumulative_variance
})

display(variance_table)

# Number of components needed for 95% variance
n_components = np.argmax(
    cumulative_variance >= 0.95
) + 1

print(
    "\nComponents needed to retain 95% variance:",
    n_components
)
```

## 7.7 Explained Variance Visualization

A scree plot is used to visualize the explained variance of each principal component.

A cumulative variance plot helps identify the number of components needed to retain a desired percentage of information.

```python
components = range(1, len(explained_variance) + 1)

plt.figure(figsize=(10, 5))

plt.plot(
    components,
    explained_variance,
    marker="o",
    label="Individual Variance"
)

plt.plot(
    components,
    cumulative_variance,
    marker="s",
    label="Cumulative Variance"
)

plt.axhline(
    y=0.95,
    linestyle="--",
    label="95% Variance"
)

plt.xlabel("Principal Components")
plt.ylabel("Explained Variance Ratio")
plt.title("PCA Explained Variance")
plt.xticks(list(components))
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

## 7.8 Dimensionality Reduction

The original seven personality features are reduced to the number of principal components required to retain at least 95% of the variance.

```python
# Apply PCA with selected components
pca_reduced = PCA(
    n_components=n_components
)

X_reduced = pca_reduced.fit_transform(
    X_scaled
)

# Create reduced dataset
pca_columns = [
    f"PC{i+1}" for i in range(n_components)
]

df_pca = pd.DataFrame(
    X_reduced,
    columns=pca_columns
)

print("Original Dataset Shape:", X.shape)
print("Reduced Dataset Shape:", df_pca.shape)

display(df_pca.head())
```

## 7.9 Two-Dimensional PCA Visualization

The first two principal components are plotted in a scatter plot to visualize the distribution of observations in reduced dimensions.

Gender is used to distinguish observations in the visualization.

```python
# Use the first two components from full PCA
pca_2d = X_pca[:, :2]

pca_plot = pd.DataFrame(
    pca_2d,
    columns=["PC1", "PC2"]
)

# Add gender labels
pca_plot["Gender"] = df_clean["Gender"].map({
    -0.48246: "Male",
     0.48246: "Female"
}).values

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=pca_plot,
    x="PC1",
    y="PC2",
    hue="Gender",
    alpha=0.7
)

plt.title("PCA - Two Dimensional Visualization")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend(title="Gender")
plt.tight_layout()
plt.show()
```

## 7.10 Experiment Conclusion

Dimensionality reduction was performed on the personality features of the drug consumption dataset.

Variance Thresholding was used to identify low-variance features. Correlation analysis and VIF were used to examine relationships and multicollinearity. Principal Component Analysis was applied to transform the original features into principal components.

Explained variance and PCA scatter plots were used to visualize the results. This process helped understand how dimensionality reduction can simplify data while retaining important information.
